```python
import hicstraw
import numpy as np
import math
import os

os.chdir('../..')
from load import generate_coverage_profile_intra_inter, load_GC_comtent, getmappability, load_num_res_site, load_cnvpytpr_to_bin, load_freec_to_bin, load_seg_to_bin
from binlib import output_bin
from metric import calculate_correlation_with_groundtruth, calculate_precision_racall, calculate_dict_correlation
from convert import convert_bin_to_bw

# back to current filder
os.chdir('Exps/exp5_apply_control')

ref_genome_folder = '../../Data/ref_genome/hg19'
exclude_chr = ['chrM', 'chrY']
resolution = 50000


# load ref genome length info
chr_length = dict()
with open(ref_genome_folder + '/hg19.len', 'r') as f:
    lines = f.read().splitlines()
    for line in lines:
        chr_name , len_ = line.split('\t')[:2]
        if chr_name in exclude_chr:
            continue
        chr_length[chr_name] = int(len_)
print('chr length : \n',chr_length)

# calculate bin number in each chr
chr_bin_number = dict()
for key in chr_length.keys():
    chr_bin_number[key] = math.floor(chr_length[key] / resolution) + 1
print('chr bin number : \n', chr_bin_number)

```

### load bias

```python
GC_dict = load_GC_comtent(ref_genome_folder + '/hg19_1kb_GCPercent.bed')

mappability_dict = dict()
for key in chr_length.keys() :
    mappability_dict[key] = np.array( getmappability(ref_genome_folder + '/hg19_mappability_50mer.bdg.gz', key, 1, chr_length[key], resolution) )

num_res_sites_dict = load_num_res_site(ref_genome_folder + '/hg19_HindIII_enzymeSites.txt')
```

```python
trg_contact_matrix = '../../Data/ContactMatrix/GSE63525_K562_combined.hic'
trg_coverage_profile_dict = dict()
for key in chr_length.keys():
    print(f'Parsing contacts between {key} and other')
    trg_coverage_profile_dict[key] = generate_coverage_profile_intra_inter(trg_contact_matrix, trg_chr = key, resolution = resolution)
```

```python
control_contact_matrix = '../../Data/ContactMatrix/GSE63525_GM12878_insitu_primary.hic'
control_coverage_profile_dict = dict()
for key in chr_length.keys():
    print(f'Parsing contacts between {key} and other')
    control_coverage_profile_dict[key] = generate_coverage_profile_intra_inter(control_contact_matrix, trg_chr = key, resolution = resolution)
```

```python
# write to file
f = open('bias_with_control_allchroms_nonzeros.txt','w+')
f.write(f'chrom\tbin_num\tcontrol\tcutSitesNumber\tgcper\tmapscore\trowsum\n')
for key in chr_bin_number.keys():
    for i in range(chr_bin_number[key]):
        control = control_coverage_profile_dict[key][i]
        num_res_sites = num_res_sites_dict[key][i]
        mappability = mappability_dict[key][i]
        GC = GC_dict[key][i]
        rowsum = trg_coverage_profile_dict[key][i]
        if num_res_sites == mappability == GC ==0 :
            continue
        f.write(f'{key}\t{i}\t{control}\t{num_res_sites}\t{GC}\t{mappability}\t{rowsum}\n')
f.close()
```

### Run script 

```python
os.system('Rscript ../../GAM_with_control.R --input bias_with_control_allchroms_nonzeros.txt --log GAM_log --output residual.txt')
```

### Merge residual to orig file and output to binfile

```python
bias_rowsum = open('bias_with_control_allchroms_nonzeros.txt','r')
residual = open('residual.txt','r')

res_dict = dict()
for key in chr_bin_number.keys():
    res_dict[key] = np.zeros(chr_bin_number[key])

bias_rowsum_lines = bias_rowsum.readlines()[1:]
residual_line = residual.readlines()

for i, bias_line in enumerate(bias_rowsum_lines):
    bias_line_split = bias_line.strip().split()
    chr_ = bias_line_split[0]
    bin_num = int(bias_line_split[1])
    res = residual_line[i]
    res_dict[chr_][bin_num] = float(res)
```

```python
output_bin(res_dict, trg_output_dir = 'exp5_bin')
```

```python
# output for visualization
convert_bin_to_bw('exp5_bin', './exp5.bw')
```
