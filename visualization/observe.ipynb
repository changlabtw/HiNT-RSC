```python
import coolbox
from coolbox.api import *
coolbox.__version__
```

```python
ref_genome_folder = '../Data/ref_genome/hg19'
exclude_chr = ['chrM', 'chrY']

# load ref genome length info
chr_length = dict()
with open(ref_genome_folder + '/hg19.len', 'r') as f:
    lines = f.read().splitlines()
    for line in lines:
        chr_name , len_ = line.split('\t')[:2]
        if chr_name in exclude_chr:
            continue
        chr_length[chr_name] = int(len_)
print('chr length : \n',chr_length)
```

```python
frame = XAxis() + \
                BedGraph(f'CNV/hic_residual.bw')+ Color("#1ab9e0") + TrackHeight(5) + Title("HiNT residual") + MinValue(-2) + MaxValue(2) +\
                BedGraph(f'CNV/hic_log2_copy_ratio.bw')+ Color("#1ab9e0") + TrackHeight(5) + Title("HiNT log2 copy ratio") + MinValue(-2) + MaxValue(2) +\
                BedGraph(f'CNV/WGS_log2_copy_ratio.bw')+ Color("#F05850") + TrackHeight(5) + Title("BICseq2 log2 copy ratio") + MinValue(-2) + MaxValue(2)  
frame.plot('chr1:0-249250621')
```
