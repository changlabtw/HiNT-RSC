```python
import coolbox
from coolbox.api import *
coolbox.__version__
```

```python
ref_genome_folder = '../Data/ref_genome/hg19'
exclude_chr = ['chrM', 'chrY']

# load ref genome length info
chr_length = dict()
with open(ref_genome_folder + '/hg19.len', 'r') as f:
    lines = f.read().splitlines()
    for line in lines:
        chr_name , len_ = line.split('\t')[:2]
        if chr_name in exclude_chr:
            continue
        chr_length[chr_name] = int(len_)
print('chr length : \n',chr_length)
```

```python
frame = XAxis() + HiCMat(f"../Data/ContactMatrix/GSE63525_K562_combined.hic", style = 'matrix') 
chr_ = 'chr1'
frame.plot(f'{chr_}:0-{chr_length[chr_]}')
```

```python
src_dir = 'CNV/exp3_pool'
win_size_list = [3,5,11,21,41]

with Color("#1ab9e0"):
    with TrackHeight(5):
        with MinValue(-2):
            with MaxValue(2):
                res_tracks = BedGraph(f'CNV/hic_residual.bw') + Title("HiNT residual") 
                for win_size in win_size_list:
                    res_tracks = res_tracks + BedGraph(f'{src_dir}/hic_residual_pool_win{win_size}.bw') + Title(f"HiNT residual(pooling with window size {win_size})") 


frame = XAxis() + HiCMat(f"../Data/ContactMatrix/GSE63525_K562_combined.hic", style='window') + Title("HI-C (K562)") + \
                BigWig(f'TAD/GSE63525_K562_50k_IS_150kb.bigwig')+ Color("#BEE067") + TrackHeight(5) + Title("insulation score (150kb)") + MinValue(-1) + MaxValue(1) +\
                res_tracks +\
                BedGraph(f'CNV/WGS_log2_copy_ratio.bw')+ Color("#F05850") + TrackHeight(5) + Title("BICseq2 log2 copy ratio") + MinValue(-2) + MaxValue(2)  
```
