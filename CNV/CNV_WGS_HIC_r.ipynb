```python
import os
import numpy as np
import pandas as pd
import math
from scipy import stats

from binlib import load_seg_file

ref_genome_folder = '../Data/ref_genome/hg19'
exclude_chr = ['chrM', 'chrY']
resolution = 50000


# load ref genome length info
chr_length = dict()
with open(ref_genome_folder + '/hg19.len', 'r') as f:
    lines = f.read().splitlines()
    for line in lines:
        chr_name , len_ = line.split('\t')[:2]
        if chr_name in exclude_chr:
            continue
        chr_length[chr_name] = int(len_)
print('chr length : \n',chr_length)

# calculate bin number in each chr
chr_bin_number = dict()
for key in chr_length.keys():
    chr_bin_number[key] = math.floor(chr_length[key] / resolution) + 1
print('chr bin number : \n', chr_bin_number)
```

### load hic CNV

```python
hic_cnv_file = 'K562_HIC_CNV/HiNTcnv_OUTPUT_K562/segmentation/K562_CNV_segments.txt'

hic_cnv = load_seg_file(hic_cnv_file, has_p_value = True)
```

### load WGS CNV with different labmda value and calculate r

```python
lambda_list = [2,3,4,5,10,30,50]

for lambda_ in lambda_list:
    wgs_cnv_file = f'K562_WGS_CNV/K562_WGS_CNV_lambda{lambda_}.txt'
    wgs_cnv = load_seg_file(wgs_cnv_file, has_p_value = False)

    hic_cnv_overall, wgs_cnv_overall = [], []
    for key in chr_bin_number.keys():
        hic_cnv_overall = hic_cnv_overall + list(hic_cnv[key])
        wgs_cnv_overall = wgs_cnv_overall + list(wgs_cnv[key])
    
    print(f'Pearson r between CNV from WGS and HIC ( lambda {lambda_} ): ', round(stats.pearsonr(hic_cnv_overall, wgs_cnv_overall)[0],3))
    print(f'Spearman r between CNV from WGS and HIC ( lambda {lambda_} ) : ', round(stats.spearmanr(hic_cnv_overall, wgs_cnv_overall)[0],3))
    print('\n')
```
